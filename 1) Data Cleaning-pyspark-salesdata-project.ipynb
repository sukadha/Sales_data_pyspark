
# Data Cleaning

### STEP 1: Remove Duplicate Records

```python
df.dropDuplicates()
```

### STEP 2: Handle NULL Values


### Drop unwanted columns

#### Check NULL counts

```python
from pyspark.sql.functions import col

df.select([col(c).isNull().alias(c) for c in df.columns]).display()
```

### Drop rows with critical NULLs

Removes rows where any of these important columns are NULL

```python
df = df.dropna(subset=["Order ID", "Product", "Sales"]).display()
```

# STEP 3: Remove Invalid / Negative Values

```python
df = df.filter(
    (col("Quantity Ordered") > 0) &
    (col("Price Each") > 0) &
    (col("Sales") > 0)
)
```

# STEP 4: Recalculate Sales (Data Consistency)

```python
df = df.withColumn(
    "Sales",
    round(col("Quantity Ordered") * col("Price Each"), 2)
)
```

# STEP 5: Clean Text Columns

# Standardize text case

```python
from pyspark.sql.functions import upper

df = df.withColumn("City", upper(col("City")))
```

# STEP 6: Validate Date & Time Columns

Remove invalid timestamps.

```python
df = df.filter(col("Order Date").isNotNull())
```

#STEP 7: Validate Derived Columns

Month must be 1–12, Hour must be 0–23.

```python
df = df.filter(
    (col("Month").between(1, 12)) &
    (col("Hour").between(0, 23))
)
```

# STEP 8: Final Schema Validation

```python
df.printSchema()
df.describe().display()
```

# Count nulls per column

If output is 0, your cleaning worked perfectly.

```python
from pyspark.sql.functions import col, sum

df.select([
    sum(col(c).isNull().cast("int")).alias(c)
    for c in df.columns
]).display()
```

# business insights

### STEP 1: Basic Business KPIs (Must-have)

Total Revenue

```python
from pyspark.sql.functions import sum

df.select(sum("Sales").alias("Total_Revenue")).display()
```

Total Orders

```python
df.selectExpr("count(distinct `Order ID`) as Total_Orders").display()
```

Average Order Value

```python
df.selectExpr("sum(Sales) / count(distinct `Order ID`) as Avg_Order_Value").display()
```

### Step 2 -Time-Based Analysis (High Impact)

Monthly Sales Trend

```python
df.groupBy("Month") \
  .sum("Sales") \
  .orderBy("Month") \
  .display()
```

Hourly Sales Trend (Peak Time)

```python
df.groupBy("Hour") \
  .sum("Sales") \
  .orderBy("Hour") \
  .display()
```

### STEP 3: Product Performance Analysis

Top 10 Products by Revenue

```python
df.groupBy("Product") \
  .sum("Sales") \
  .orderBy("sum(Sales)", ascending=False) \
  .limit(10) \
  .display()
```

```python
df.groupBy("Product") \
  .sum("Sales") \
  .sort(desc("sum(Sales)")) \
  .limit(10) \
  .display()
```

Most Sold Products (Quantity)

```python
df.groupBy("Product") \
  .sum("Quantity Ordered") \
  .orderBy("sum(Quantity Ordered)", ascending=False) \
  .display()
```

### STEP 4: Location-Based Insights

City-wise Revenue

```python
df.groupBy("City") \
  .sum("Sales") \
  .orderBy("sum(Sales)", ascending=False) \
  .display()
```

Best City per Month

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import rank, col

windowSpec = Window.partitionBy("Month").orderBy(col("Sales").desc())

df.withColumn("rank", rank().over(windowSpec)) \
  .filter(col("rank") == 1) \
  .display()
```

### STEP 7: Advanced Analytics 

Product Pair Analysis (Market Basket)

```python
df.groupBy("Order ID") \
  .agg(collect_set("Product").alias("Products")) \
  .filter(size("Products") > 1) \
  .display()
```

### STEP 6: Pivot Analysis (Excel-like)

Product vs Month Sales

```python
df.groupBy("Product") \
  .pivot("Month") \
  .sum("Sales") \
  .display()
  
```

#----------------------VISUALIZATION----------------

### Monthly Sales → Line Chart

```python
monthly_sales = df.groupBy("Month") \
    .agg(round(sum("Sales"), 2).alias("Total_Sales")) \
    .orderBy("Month")

monthly_sales.display()
```

### Hourly Sales → Bar Chart

```python
hourly_sales = df.groupBy("Hour") \
    .agg(round(sum("Sales"), 2).alias("Hourly_Sales")) \
    .orderBy("Hour")

hourly_sales.display()
```

### City Revenue → Bar Chart

```python
city_sales = df.groupBy("City") \
    .agg(round(sum("Sales"), 2).alias("City_Revenue")) \
    .orderBy(desc("City_Revenue"))

city_sales.display()
```

###Top Products → Horizontal Bar Chart

```python
top_products = df.groupBy("Product") \
    .agg(sum("Quantity Ordered").alias("Total_Quantity")) \
    .orderBy(desc("Total_Quantity"))

top_products.display()
```

### Product vs Month → Pivot Heatmap

```python
product_month_sales = df.groupBy("Product") \
    .pivot("Month") \
    .agg(round(sum("Sales"), 2))

product_month_sales.display()
```
